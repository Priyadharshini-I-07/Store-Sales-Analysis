```python
import pandas as pd
df=pd.read_csv("stores_sales_forecasting.csv",encoding="latin1")
df.drop("Sales",axis=1,inplace=False)
print(df)
```

```python
grouped_df=df.groupby('Ship Date')['Profit'].mean()
print(grouped_df)
```

```python
df["Q_Status"]=df["Quantity"].apply(lambda x:'High' if x>=5 else 'Normal')
print(df['Q_Status'].head(10))
print(df['Q_Status'].tail(10))
```

```python
def Q_Status(x):
    if x>=5:
        return "High"
    else:
        return "Normal"

df["Q_Status"]=df["Quantity"].apply(Q_Status)
print(df['Q_Status'].head(10))
print(df['Q_Status'].tail(10))
```
